```python
import pandas as pd
import joblib
df = pd.read_csv("original_data.csv")
```

```python
df.columns
```

```python
from sklearn.preprocessing import LabelEncoder
import numpy as np


label_cols = ['gender', 'area', 'affected_eye','pterygium_location','surgery_type','followup1_complication', 'followup2_complication']

label_encoders = {}
for col in label_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
    joblib.dump(le, f"label_encoder_{col}.pkl")



df[label_cols].head()
```

```python
df.to_csv("prep_data.csv",index=False)
```
